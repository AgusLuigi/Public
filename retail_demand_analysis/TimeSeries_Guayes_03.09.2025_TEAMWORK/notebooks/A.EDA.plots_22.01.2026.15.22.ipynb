```python
import numpy as np
from Favorita_TSA.preprocess_eda import load_table

from Favorita_TSA.utils.data_loader import df_to_parquet
from Favorita_TSA.utils.dataset import PreDataset


def evaluate_forecast_reliability():
    """
    Optimierte Berechnung der Forecast-Eignung mit zusätzlichem Zeitreihen-Plot.
    """
    # 1. Daten laden (RAM-schonend)
    df_TEMP = load_table(PreDataset.STORE_ITEM_DAILY)
    df_item_global = load_table(PreDataset.ITEM_DAILY)

    if not pd.api.types.is_datetime64_any_dtype(df_TEMP["date"]):
        df_TEMP["date"] = pd.to_datetime(df_TEMP["date"])

    # --- Schritt A: Globale Signalstärke (Vektorisiert) ---
    item_stats = (
        df_item_global.groupby("item_nbr")
        .agg(
            global_cv=(
                "unit_sales_sum",
                lambda x: x.std() / x.mean() if x.mean() > 0 else np.nan,
            ),
            global_mean=("unit_sales_sum", "mean"),
        )
        .reset_index()
    )

    # --- Schritt B: Store-Item Metriken (Vektorisiert) ---
    print("Berechne Metriken für Store-Item Kombinationen...")
    df_metrics = (
        df_TEMP.groupby(["store_nbr", "item_nbr"])
        .agg(
            first_date=("date", "min"),
            last_date=("date", "max"),
            obs_days=("unit_sales_sum", "count"),
            mean_sales=("unit_sales_sum", "mean"),
            std_sales=("unit_sales_sum", "std"),
            zero_count=(
                "unit_sales_sum",
                lambda x: (x == 0).sum(),
            ),  # zero count muss anders berechnet werden weil wir keine 0 einträge haben
        )
        .reset_index()
    )

    # Berechnete Spalten
    df_metrics["history_days"] = (
        df_metrics["last_date"] - df_metrics["first_date"]
    ).dt.days + 1
    df_metrics["coverage_ratio"] = df_metrics["obs_days"] / df_metrics["history_days"]
    df_metrics["cv_local"] = df_metrics["std_sales"] / df_metrics["mean_sales"]
    df_metrics["zero_ratio"] = df_metrics["zero_count"] / df_metrics["obs_days"]

    # Merge mit globalen Stats
    df_metrics = df_metrics.merge(item_stats, on="item_nbr", how="left")

    # Finaler Filter (Zuverlässigkeits-Score)
    df_metrics["is_reliable"] = (
        (df_metrics["history_days"] > 365)
        & (df_metrics["cv_local"] < 1.0)
        & (df_metrics["coverage_ratio"] > 0.8)
    )

    # --- Schritt C: Visualisierung 1 (Scatter Plot: Signal vs. Rauschen) ---
    plot_df = df_metrics[df_metrics["cv_local"] < 3.0].copy()
    fig1 = px.scatter(
        plot_df,
        x="mean_sales",
        y="cv_local",
        color="is_reliable",
        size="history_days",
        hover_data=["store_nbr", "item_nbr"],
        title="Zuverlässigkeits-Analyse: Rauschen (CV) vs. Volumen",
        labels={"mean_sales": "Durchschnittlicher Absatz", "cv_local": "Rauschen (CV)"},
        color_discrete_map={True: "#2ECC71", False: "#E74C3C"},
        template="plotly_dark",
    )
    fig1.add_hline(y=1.0, line_dash="dash", line_color="white")
    fig1.show()

    # --- Schritt D: NEU - Visualisierung 2 (Zeitverlauf der Verkäufe) ---
    print("Erstelle Zeitverlauf-Plot...")
    # Wir mergen die Zuverlässigkeits-Info zurück in die Daily-Daten für den Plot
    df_plot_time = df_TEMP.merge(
        df_metrics[["store_nbr", "item_nbr", "is_reliable"]],
        on=["store_nbr", "item_nbr"],
    )

    # Für eine saubere Darstellung aggregieren wir auf die Verlässlichkeit pro Tag
    df_time_series = (
        df_plot_time.groupby(["date", "is_reliable"])["unit_sales_sum"]
        .sum()
        .reset_index()
    )

    fig2 = px.line(
        df_time_series,
        x="date",
        y="unit_sales_sum",
        color="is_reliable",
        title="Verkaufte Einheiten über Zeit (Gefiltert nach Forecast-Zuverlässigkeit)",
        labels={
            "unit_sales_sum": "Summe Verkaufte Einheiten",
            "date": "Datum",
            "is_reliable": "Zuverlässig (LSTM-bereit)",
        },
        color_discrete_map={True: "#2ECC71", False: "#E74C3C"},
        template="plotly_dark",
    )
    fig2.show()

    # --- Schritt E: Numerische Tabellenausgabe ---
    summary = (
        df_metrics.groupby("is_reliable")
        .agg(
            Anzahl_Kombinationen=("item_nbr", "count"),
            Avg_CV=("cv_local", "mean"),
            Avg_Sales=("mean_sales", "mean"),
            Gesamt_Absatz=("mean_sales", lambda x: x.sum() * 365),  # Schätzung pro Jahr
        )
        .round(2)
    )

    print("\n" + "=" * 60)
    print("ZUSAMMENFASSUNG: FORECAST ZUVERLÄSSIGKEIT & SIGNALSTÄRKE")
    print("=" * 60)
    print(summary)
    print("=" * 60 + "\n")

    # Speichern als Packet-Typ [2026-01-07]
    output_path = "data/processed/preprocessed/forecast_reliability.parquet"
    df_to_parquet(df_metrics, output_path)

    return df_metrics
```

```python
from Favorita_TSA.preprocess_eda import load_table

from Favorita_TSA.utils.dataset import PreDataset


def evaluate_forecast_reliability():
    """
    Optimierte Berechnung der Forecast-Eignung mit robusten Dropdown-Menüs.
    """
    # 1. Daten laden (RAM-schonend)
    df_TEMP = load_table(PreDataset.STORE_ITEM_DAILY)
    df_item_global = load_table(PreDataset.ITEM_DAILY)

    if not pd.api.types.is_datetime64_any_dtype(df_TEMP["date"]):
        df_TEMP["date"] = pd.to_datetime(df_TEMP["date"])

    # --- Schritt A: Globale Signalstärke ---
    item_stats = (
        df_item_global.groupby("item_nbr")
        .agg(
            global_cv=(
                "unit_sales_sum",
                lambda x: x.std() / x.mean() if x.mean() > 0 else np.nan,
            ),
            global_mean=("unit_sales_sum", "mean"),
        )
        .reset_index()
    )

    # --- Schritt B: Store-Item Metriken ---
    print("Berechne Metriken für Store-Item Kombinationen...")
    df_metrics = (
        df_TEMP.groupby(["store_nbr", "item_nbr"])
        .agg(
            first_date=("date", "min"),
            last_date=("date", "max"),
            obs_days=("unit_sales_sum", "count"),
            mean_sales=("unit_sales_sum", "mean"),
            std_sales=("unit_sales_sum", "std"),
        )
        .reset_index()
    )

    df_metrics["history_days"] = (
        df_metrics["last_date"] - df_metrics["first_date"]
    ).dt.days + 1
    df_metrics["coverage_ratio"] = df_metrics["obs_days"] / df_metrics["history_days"]
    df_metrics["cv_local"] = df_metrics["std_sales"] / df_metrics["mean_sales"]

    df_metrics = df_metrics.merge(item_stats, on="item_nbr", how="left")

    # Zuverlässigkeits-Logik
    df_metrics["is_reliable"] = (
        (df_metrics["history_days"] > 365)
        & (df_metrics["cv_local"] < 1.0)
        & (df_metrics["coverage_ratio"] > 0.8)
    )

    # --- Schritt C: Visualisierung 1 (Scatter Plot) ---
    plot_df = df_metrics[df_metrics["cv_local"] < 3.0].copy()
    stores = sorted(plot_df["store_nbr"].unique())
    fig1 = go.Figure()

    for store in stores:
        store_data = plot_df[plot_df["store_nbr"] == store]
        fig1.add_trace(
            go.Scatter(
                x=store_data["mean_sales"],
                y=store_data["cv_local"],
                mode="markers",
                name=f"Store {store}",
                visible=True,
                marker=dict(
                    size=np.clip(
                        store_data["history_days"] / 100, 4, 15
                    ),  # Begrenzte Größe für Performance
                    color=store_data["is_reliable"].map(
                        {True: "#2ECC71", False: "#E74C3C"}
                    ),
                    line=dict(width=0.5, color="white"),
                ),
                customdata=store_data[["store_nbr", "item_nbr"]],
                hovertemplate="Store: %{customdata[0]}<br>Item: %{customdata[1]}<br>CV: %{y:.2f}<extra></extra>",
            )
        )

    dropdown_c = [
        dict(
            label="Alle Stores",
            method="update",
            args=[{"visible": [True] * len(stores)}],
        )
    ]
    for i, s in enumerate(stores):
        v = [False] * len(stores)
        v[i] = True
        dropdown_c.append(
            dict(label=f"Store {s}", method="update", args=[{"visible": v}])
        )

    fig1.update_layout(
        updatemenus=[{"buttons": dropdown_c, "x": 0, "y": 1.15, "xanchor": "left"}],
        template="plotly_dark",
        title="Rauschen vs. Volumen",
    )
    fig1.show()

    # --- Schritt D: Zeitverlauf (Stabile Trace-Logik) ---
    print("Erstelle Zeitverlauf-Plot...")
    df_time = df_TEMP.merge(
        df_metrics[["store_nbr", "item_nbr", "is_reliable"]],
        on=["store_nbr", "item_nbr"],
    )
    df_ts_agg = (
        df_time.groupby(["date", "store_nbr", "is_reliable"])["unit_sales_sum"]
        .sum()
        .reset_index()
    )

    fig2 = go.Figure()
    trace_map = []  # Speichert, welche Traces zu welchem Store gehören

    current_idx = 0
    for store in stores:
        indices_for_this_store = []
        for status, color, label in [
            (True, "#2ECC71", "Reliable"),
            (False, "#E74C3C", "Unstable"),
        ]:
            subset = df_ts_agg[
                (df_ts_agg["store_nbr"] == store) & (df_ts_agg["is_reliable"] == status)
            ]
            fig2.add_trace(
                go.Scatter(
                    x=subset["date"],
                    y=subset["unit_sales_sum"],
                    mode="lines",
                    name=f"S{store}-{label}",
                    line=dict(color=color),
                    visible=True,
                )
            )
            indices_for_this_store.append(current_idx)
            current_idx += 1
        trace_map.append(indices_for_this_store)

    dropdown_d = [
        dict(
            label="Alle Stores",
            method="update",
            args=[{"visible": [True] * current_idx}],
        )
    ]
    for i, store in enumerate(stores):
        v = [False] * current_idx
        for idx in trace_map[i]:
            v[idx] = True
        dropdown_d.append(
            dict(label=f"Store {store}", method="update", args=[{"visible": v}])
        )

    fig2.update_layout(
        updatemenus=[{"buttons": dropdown_d, "x": 0, "y": 1.15, "xanchor": "left"}],
        template="plotly_dark",
        title="Zeitverlauf nach Zuverlässigkeit",
    )
    fig2.show()

    # --- Schritt E: Numerische Tabelle [2025-10-29] ---
    summary = (
        df_metrics.groupby("is_reliable")
        .agg(
            Anzahl=("item_nbr", "count"),
            Avg_CV=("cv_local", "mean"),
            Avg_Sales=("mean_sales", "mean"),
        )
        .round(2)
    )
    print("\nSUMMARY:\n", summary)

    # Speichern als Packet-Typ [2026-01-07]
    df_to_parquet(
        df_metrics, "data/processed/preprocessed/forecast_reliability.parquet"
    )
    return df_metrics
```

```python
evaluate_forecast_reliability()
```

- Erklärung der ErgebnisseSignalstärke ($CV$): Die Y-Achse misst das Rauschen. Ein $CV < 1.0$ (unter der gestrichelten Linie) zeigt ein stabiles Signal, das ideal für dein LSTM-Modell ist.Volumen ($Mean Sales$): Rechts im Plot befinden sich die Umsatzträger. Wenn ein Item rechts unten (Grün) liegt, ist es ein "Champion" für den Forecast.Aggregations-Eignung: Wenn agg_benefit hoch ist, ist der lokale Store-Absatz zu erratisch. In diesem Fall ist es besser, auf globaler Ebene zu trainieren.Zuverlässigkeit: Nur Items mit is_reliable = True sollten für ein individuelles Store-Item-Modell genutzt werden, um Overfitting auf Rauschen zu vermeiden.
